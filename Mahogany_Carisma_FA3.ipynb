## 📦 Part 1: What is an Array / List?

Imagine you are managing an online game inventory, a class record, or a music playlist.
- **Without Arrays:** You would need separate variables like `item1 = "Notebook"`, `item2 = "Pen"`, `item3 = "Calculator"`, etc. This gets messy very quickly!
- **With Arrays / Lists:** You can group all related items into **one single variable**:
  ```python
  backpack = ["Notebook", "Pen", "Calculator", "Laptop", "Book"]
  ```

### 💡 Real-World Analogy: A Row of Boxes
Think of a list as a row of numbered boxes sitting on a shelf. Each box holds **one item**, and each box has a position number called an **index**.

Let's run the code cell below to create our first Python lists!

````python
# Creating our first Python Lists (Datasets from Study Guide)

# Dataset A: School Backpack Items (Strings)
backpack = ["Notebook", "Pen", "Calculator", "Laptop", "Book"]

# Dataset B: Class Quiz Scores (Numbers)
quiz_scores = [85, 90, 78, 92, 88]

# Dataset C: Favorite Foods (Strings)
favorite_foods = ["Pizza", "Burger", "Pasta", "Fried Chicken", "Ice Cream"]

print("Backpack Items:", backpack)
print("Quiz Scores:", quiz_scores)
print("Favorite Foods:", favorite_foods)
````

## 🔢 Part 2: Understanding 0-Based Indexing

In Python (and most programming languages like C++, Java, and JavaScript), **indexes always start counting at 0**, NOT 1!

| Index Position | `backpack` Item | `quiz_scores` Value |
| :---: | :---: | :---: |
| **`0`** | `"Notebook"` | `85` |
| **`1`** | `"Pen"` | `90` |
| **`2`** | `"Calculator"` | `78` |
| **`3`** | `"Laptop"` | `92` |
| **`4`** | `"Book"` | `88` |

### 🛠️ Accessing and Modifying Items with Square Brackets `[ ]`
- **To access an item:** `list_name[index]`
- **To modify an item:** `list_name[index] = new_value`

Run the code below to see direct indexing in action!

````python
# Activity 1: Direct Index Accessing

#Create your list here
list_name = ['Frances', 'Terence', 'Tohan', 'Hannah', 'Seth', 'Zia']


#access your list here
print("--- Direct Index Access ---")
print("Element at index 0:", list_name[0])
print("Element at index 1:", list_name[1])
print("Element at index 2:", list_name[2])
print("Element at index 3:", list_name[3])
print("Element at index 4:", list_name[4])
print('Element at index 5:', list_name[5])

print("\n--- Modifying an Element via Index ---")
print("Original index 2 item:",list_name[2])

# Replacing index 2 __________ with new Value
list_name[2] = list_name[3]

print("Updated index 2 item:", list_name[2])
print("Updated  List:", list_name)
````

## 🔄 Part 3: Array Traversal Using Loops

While direct indexing is great for single items, **traversal** means visiting every element in an array one by one.

We use `for` loops to make traversal fast, clean, and efficient!

### Two Ways to Loop Through a List:
1. **Looping by Item:** `for item in list_name:`
2. **Looping by Index:** `for i in range(len(list_name)):`

Let's test both methods below!

````python
# Array Traversal Examples

print("=== Method 1: Looping by Item ===")
for item in backpack:
    print(f"🎒 Item: {item}")

print("\n=== Method 2: Looping by Index (with Position Labels) ===")
for i in range(len(backpack)):
    print(f"Index {i} -> {backpack[i]}")

print("\n=== Processing Data During Traversal (Calculating Average Score) ===")
total_score = 0
for score in quiz_scores:
    total_score += score

average_score = total_score / len(quiz_scores)
print(f"Total Points: {total_score}")
print(f"Average Score: {average_score:.2f}")
````

## ✂️ Part 4: List Slicing & Searching

### 1. List Slicing (`list[start:end]`)
Slicing allows you to extract a sub-list (a portion of your list).
- Syntax: `list_name[start:end]`
- **Remember:** The slice starts at `start` index and goes up to, but **does NOT include**, `end` index!
- Example: `backpack[1:4]` gets items at index 1, 2, and 3.

### 2. Linear Searching
Searching means checking if a target value exists in your list and finding its index position.

````python
# Slicing and Searching Demo (Matching Activity 2 Requirements)

# Reset list
backpack = ["Notebook", "Pen", "Tablet", "Laptop", "Book"]

print("Original Backpack List:", backpack)

# Extracting a sliced portion from index 1 to 4 (indexes 1, 2, 3)
sliced_backpack = backpack[1:4]
print("Sliced Sub-List (1:4):", sliced_backpack)

# Linear Search Function
search_item = "Laptop"

if search_item in sliced_backpack:
    found_index = sliced_backpack.index(search_item)
    print(f"🔍 Search Result: '{search_item}' was found in the sliced list at sub-index {found_index}!")
else:
    print(f"🔍 Search Result: '{search_item}' was not found in the sliced list.")
````

## 🧰 Part 5: Essential Python Built-in List Methods

Python provides powerful built-in methods and functions to modify and inspect lists without writing long loops!

| Method / Function | What It Does | Example |
| :--- | :--- | :--- |
| **`len(list)`** | Returns total number of items | `len(backpack)` |
| **`sum(list)`** | Calculates sum of numeric items | `sum(quiz_scores)` |
| **`max(list)` / `min(list)`** | Finds highest / lowest value | `max(quiz_scores)` |
| **`append(item)`** | Adds item to the **end** of list | `backpack.append("Binder")` |
| **`insert(index, item)`** | Inserts item at specific position | `backpack.insert(1, "Ruler")` |
| **`remove(item)`** | Removes first occurrence of item | `backpack.remove("Pen")` |
| **`pop(index)`** | Removes & returns item at index | `backpack.pop(0)` |
| **`sort()` / `sorted()`** | Sorts list in ascending order | `quiz_scores.sort()` |
| **`reverse()`** | Reverses order of list items | `backpack.reverse()` |

Try running the code below to see these methods in action!

````python
# Demonstrating Built-in Methods

scores = [85, 90, 78, 92, 88]
print("Initial Scores:", scores)

# Math Functions
print(f"Total Items: {len(scores)}")
print(f"Sum of Scores: {sum(scores)}")
print(f"Highest Score: {max(scores)}")
print(f"Lowest Score: {min(scores)}")

# Modifying List
scores.append(95)
print("After append(95):", scores)

scores.sort()
print("After sort():", scores)

scores.reverse()
print("After reverse():", scores)
````

## 🏆 Part 7: Mini-Project — Student Score Analyzer Mini-App

Create a complete modular text-based program that manages and analyzes student scores!

This program uses a **modular structure with functions** and a clear **interactive menu** matching Activity 3 (Data Analysis Mini-App).

Run the cell below to test the Student Score Analyzer!

````python
# Student Score Analyzer Mini-App

scores_data = [85, 90, 78, 92, 88]

def display_scores(scores):
    print("\n--- Current Student Scores ---")
    for i, s in enumerate(scores):
        print(f"Student #{i+1} (Index {i}): {s}")

def search_score(scores, target):
    if target in scores:
        idx = scores.index(target)
        print(f"✅ Score {target} found at Student #{idx+1} (Index {idx}).")
    else:
        print(f"❌ Score {target} not found in records.")

def compute_statistics(scores):
    if not scores:
        print("No scores available.")
        return
    avg = sum(scores) / len(scores)
    print(f"\n--- Score Statistics ---")
    print(f"Total Students: {len(scores)}")
    print(f"Average Score:  {avg:.2f}")
    print(f"Highest Score:  {max(scores)}")
    print(f"Lowest Score:   {min(scores)}")

def update_score(scores, index, new_score):
    if 0 <= index < len(scores):
        old_val = scores[index]
        scores[index] = new_score
        print(f"✅ Updated Index {index}: {old_val} -> {new_score}")
    else:
        print(f"❌ Invalid index! Must be between 0 and {len(scores)-1}.")

def run_mini_app():
    print("=== Welcome to Student Score Analyzer ===")

    # Demonstration Run
    display_scores(scores_data)
    compute_statistics(scores_data)

    print("\n--- Performing Search ---")
    search_score(scores_data, 92)

    print("\n--- Updating Score ---")
    update_score(scores_data, 2, 82) # Update 78 to 82

    display_scores(scores_data)
    compute_statistics(scores_data)

# Run the Mini-App
run_mini_app()
````
